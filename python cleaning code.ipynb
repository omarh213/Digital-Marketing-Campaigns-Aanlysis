```python
df["CVR"]=df["conversions"]/df["clicks"]
```

```python
df["CPC"]=df["spend"]/df["clicks"]
```

```python
df["revenue"]=df["Profit"]-df["spend"]
```

```python
df["RPC"]=df["revenue"]/df["clicks"]
```

```python
df["CVR"] = (df["conversions"] / df["clicks"]).fillna(0)
df["CPC"] = (df["spend"] / df["clicks"]).fillna(0)
df["RPC"] = (df["revenue"] / df["clicks"]).fillna(0)
```

```python
df.sort_values(by="date",inplace=True)
```

```python
df.reset_index(drop=True,inplace=True)
```

```python
df["Month"]=df["date"].dt.month_name()
```

```python
df["Year"]=df["date"].dt.year
```

```python
df["ROI"]=(df["spend"]/df["Profit"])*100
```

```python
df.to_excel("Marketing_Cleaned.xlsx", index=False)
```
